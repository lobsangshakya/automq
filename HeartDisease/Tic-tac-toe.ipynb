```python
import random

def print_board(b):
    print("\n")
    for i in range(0, 9, 3):
        print(" ", b[i], "|", b[i+1], "|", b[i+2])
        if i < 6:
            print("-------------")
    print("\n")

def winner(b):
    wins = [(0,1,2),(3,4,5),(6,7,8),
            (0,3,6),(1,4,7),(2,5,8),
            (0,4,8),(2,5,8),(0,4,8),(2,4,6)]
    for x,y,z in wins:
        if b[x]==b[y]==b[z] and b[x] != ' ':
            return b[x]
    return None

def moves(b):
    return [i for i,x in enumerate(b) if x==' ']

def make(b, m, p):
    b2 = b.copy()
    b2[m] = p
    return b2
```

```python
Q = {}
alpha, gamma, eps = 0.3, 0.9, 0.1  # learning rate, discount, exploration

def getQ(state, move):
    return Q.get((tuple(state), move), 0)

def best_move(state):
    possible = moves(state)
    if random.random() < eps:
        return random.choice(possible)
    qs = [getQ(state, m) for m in possible]
    return possible[qs.index(max(qs))]
```

```python
def train(episodes=5000):
    for _ in range(episodes):
        b = [' '] * 9
        while True:
            m = best_move(b)
            nb = make(b, m, 'X')
            w = winner(nb)
            if w or ' ' not in nb:
                Q[(tuple(b), m)] = getQ(b, m) + alpha * ((1 if w=='X' else 0) - getQ(b, m))
                break

            # Opponent random
            om = random.choice(moves(nb))
            nb2 = make(nb, om, 'O')
            w = winner(nb2)
            r = 1 if w=='X' else -1 if w=='O' else 0
            future = [getQ(nb2, a) for a in moves(nb2)] or [0]
            Q[(tuple(b), m)] = getQ(b, m) + alpha * (r + gamma * max(future) - getQ(b, m))
            if w or ' ' not in nb2:
                break
            b = nb2
    print("✅ Training complete")

train(3000)
```

```python
def play():
    b = [' '] * 9
    print_board(b)
    while True:
        # AI move
        m = best_move(b)
        b = make(b, m, 'X')
        print_board(b)
        if winner(b) or ' ' not in b:
            print("Winner:", winner(b) or "Draw!")
            break

        # Human move
        mv = int(input("Your move (0-8): "))
        while mv not in moves(b):
            mv = int(input("Invalid. Try again: "))
        b = make(b, mv, 'O')
        print_board(b)
        if winner(b) or ' ' not in b:
            print("Winner:", winner(b) or "Draw!")
            break

play()
```
